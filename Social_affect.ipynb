```python
data=pd.read_csv('Social Media DataSet.csv')
```

```python
df=pd.DataFrame(data)
```

```python
clm=df.select_dtypes(include=['object','string']).columns 
df[clm]=df[clm].apply(lambda x:(
    x.str.strip().str.replace('"',"",regex=False).str.replace("'","",regex=False).str.replace('-',"",regex=False).str.replace('-',"",regex=False).str.title()
))
```
